```python
import datetime
# fix date fields "openHouseStartDate": "2026-01-17T13:00:00" to "2026-01-17 13:00:00"
data["openHouseStartDate"] = data["openHouseStartDate"].replace("T", " ")
data["openHouseEndDate"] = data["openHouseEndDate"].replace("T", " ")

# convert to datetime objects
data["openHouseStartDate"] = datetime.datetime.strptime(data["openHouseStartDate"], "%Y-%m-%d %H:%M:%S")
data["openHouseEndDate"] = datetime.datetime.strptime(data["openHouseEndDate"], "%Y-%m-%d %H:%M:%S")

df = spark.createDataFrame(data=[data], schema=schema)
df.show(truncate=False)
df.printSchema()
```

```python
table_name = "real_estate"
jdbc_url = "jdbc:postgresql://postgres:5432/real_estate_db"
df.write.format("jdbc") \
  .option("url", jdbc_url) \
  .option("dbtable", table_name) \
  .option("user", "postgres") \
  .option("password", "postgres") \
  .option("driver", "org.postgresql.Driver") \
  .mode("overwrite") \
  .save()
```
